```python
import pandas as pd
import os

# Ruta donde están los .xlsx originales
input_dir = "../original_xlsx/"
# Ruta de salida para los .csv nuevos (sobreescribirá los viejos)
output_dir = "../data/"

# Asegúrate de que existen los archivos .xlsx con estos nombres
years = [2019, 2020, 2021, 2022, 2023]

for year in years:
    input_file = os.path.join(input_dir, f"zscore_sample{year}.xlsx")
    output_file = os.path.join(output_dir, f"zscore_sample{year}.csv")

    try:
        df = pd.read_excel(input_file, sheet_name=0, header=None)  # 👈 sin encabezado
        df.to_csv(output_file, index=False, header=False, encoding="utf-8")  # 👈 sin header
        print(f"✅ {year}: archivo convertido y guardado correctamente.")
    except Exception as e:
        print(f"❌ Error con {year}: {e}")
```

```python
from IPython.display import FileLink, display

years = [2019, 2020, 2021, 2022, 2023]
for year in years:
    file_path = f"../data/zscore_sample{year}.csv"
    if os.path.exists(file_path):
        display(FileLink(file_path))
```
